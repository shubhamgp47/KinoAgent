```python
import sqlite3
import os

OSCARS_DB_PATH = r"C:\Users\shubh\OneDrive\Documents\Tutorials\FilmGPT\FilmGPT\data\oscars_db\oscars.db"

print("Exists?", os.path.exists(OSCARS_DB_PATH))

if not os.path.exists(OSCARS_DB_PATH):
    print("Oscars database file not found")
else:
    try:
        conn = sqlite3.connect(OSCARS_DB_PATH)
        cursor = conn.execute("SELECT COUNT(*) FROM oscars_nominations;")
        count = cursor.fetchone()[0]
        conn.close()
        print("Connected OK, row count:", count)
    except Exception as e:
        print("Connection or query error:", e)
```

```python
import sqlite3


conn = sqlite3.connect(OSCARS_DB_PATH)
cursor = conn.execute("""
SELECT year_film, year_ceremony, category, canon_category,
       nominee_name, film, winner
FROM oscars_nominations
WHERE film COLLATE NOCASE = 'Everything Everywhere All At Once'
ORDER BY year_ceremony;
""")
rows = cursor.fetchall()
conn.close()

for row in rows:
    print(row)
```
